## 1. The Math Formula

Cosine similarity measures the angle between two vectors, **not** their length.

$$
\text{cosine\_similarity}(A, B) = \frac{A \cdot B}{\|A\| \times \|B\|}
$$

Where:
- $A \cdot B$ = **Dot product** of vectors A and B
- $\|A\|$ = **Magnitude (length)** of vector A = $\sqrt{a_1^2 + a_2^2 + ... + a_n^2}$
- $\|B\|$ = Magnitude of vector B

**Result range:**
- `1` → vectors point in the exact same direction (very similar)
- `0` → vectors are perpendicular (unrelated)
- `-1` → vectors point in opposite directions (opposite meaning)

### 2. Define Two Simple Vectors
We start with small, easy-to-check numbers before moving to real data.

```python
vector_A = [2, 4, 1]
vector_B = [1, 3, 5]

print("Vector A:", vector_A)
print("Vector B:", vector_B)
```

### 3. Step 1 — Dot Product (from scratch)

Formula: $A \cdot B = a_1 b_1 + a_2 b_2 + ... + a_n b_n$

```python
def dot_product(A, B):
    total = 0
    for a, b in zip(A, B):
        total += a * b
    return total

dp = dot_product(vector_A, vector_B)
print("Dot Product (A . B):", dp)
```

### 4. Step 2 — Magnitude of a Vector (from scratch)

Formula: $\|A\| = \sqrt{a_1^2 + a_2^2 + ... + a_n^2}$

```python
import math

def magnitude(A):
    total = 0
    for a in A:
        total += a ** 2
    return math.sqrt(total)

mag_A = magnitude(vector_A)
mag_B = magnitude(vector_B)

print("Magnitude of A:", round(mag_A, 4))
print("Magnitude of B:", round(mag_B, 4))
```

### 5. Step 3 — Cosine Similarity (from scratch)
Now combine both formulas.

```python
def cosine_similarity_manual(A, B):
    dp = dot_product(A, B)
    mag_A = magnitude(A)
    mag_B = magnitude(B)
    similarity = dp / (mag_A * mag_B)
    return similarity

result_manual = cosine_similarity_manual(vector_A, vector_B)
print("Cosine Similarity (Manual Calculation):", round(result_manual, 4))
```

**Discussion Point:** We just calculated cosine similarity using **only** basic Python — no libraries. This is exactly what happens internally inside NumPy or scikit-learn functions.

### 6. Verify Using NumPy
NumPy gives the same answer, but with cleaner, faster code.

```python
import numpy as np

A = np.array(vector_A)
B = np.array(vector_B)

dot = np.dot(A, B)
norm_A = np.linalg.norm(A)
norm_B = np.linalg.norm(B)

result_numpy = dot / (norm_A * norm_B)
print("Cosine Similarity (NumPy):", round(result_numpy, 4))
```

### 7. Verify Using scikit-learn
This is the function used in real-world ML pipelines.

```python
!pip install -q scikit-learn
```

```python
from sklearn.metrics.pairwise import cosine_similarity

# sklearn expects 2D arrays: shape (1, n)
A_2d = np.array(vector_A).reshape(1, -1)
B_2d = np.array(vector_B).reshape(1, -1)

result_sklearn = cosine_similarity(A_2d, B_2d)[0][0]
print("Cosine Similarity (scikit-learn):", round(result_sklearn, 4))
```

### 8. Compare All Three Results

All three methods should give the **exact same number**, proving our manual math was correct.

```python
print("Manual     :", round(result_manual, 6))
print("NumPy      :", round(result_numpy, 6))
print("scikit-learn:", round(result_sklearn, 6))

assert round(result_manual, 4) == round(result_numpy, 4) == round(result_sklearn, 4)
print("\nAll three methods match!")
```

### 9. Try More Examples (Intermediate)

Let's test edge cases to build intuition.

```python
examples = {
    "Identical vectors": ([1, 2, 3], [1, 2, 3]),
    "Opposite vectors": ([1, 2, 3], [-1, -2, -3]),
    "Perpendicular vectors": ([1, 0], [0, 1]),
    "Scaled vectors (same direction)": ([1, 2, 3], [2, 4, 6]),
}

for name, (a, b) in examples.items():
    sim = cosine_similarity_manual(a, b)
    print(f"{name:35s} -> {round(sim, 4)}")
```

**Discussion Point:**
- Identical vectors → similarity = `1`
- Opposite vectors → similarity = `-1`
- Perpendicular vectors → similarity = `0`
- Scaled vectors (same direction, different length) → still `1`, because cosine similarity ignores magnitude and only cares about **direction**. This is why it's ideal for comparing text embeddings, which can have different lengths.

### 10. Key Takeaways (Recap)

| Term | Meaning |
|---|---|
| Dot Product | Sum of element-wise multiplication of two vectors |
| Magnitude | Length of a vector, from the origin |
| Cosine Similarity | Dot product divided by product of magnitudes |
| Range | -1 (opposite) to 1 (identical direction) |
| Why it matters | Compares *direction/meaning*, ignores vector length — perfect for text embeddings |

### Exercise 1: `cosine_similarity_manual` using list comprehensions

```python
def cosine_similarity_lc(A, B):
    dp = sum(a * b for a, b in zip(A, B))
    mag_A = math.sqrt(sum([a ** 2 for a in A]))
    mag_B = math.sqrt(sum([b ** 2 for b in B]))
    return dp / (mag_A * mag_B)

print(round(cosine_similarity_lc(vector_A, vector_B), 4))
print(round(cosine_similarity_manual(vector_A, vector_B), 4))   # same value
```
