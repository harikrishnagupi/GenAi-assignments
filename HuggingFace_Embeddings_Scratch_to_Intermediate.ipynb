## 1. What is an Embedding?- An **embedding** is a list of numbers (a vector) that represents the *meaning* of a piece of text.- Similar sentences → similar (close) vectors.- Different sentences → far apart vectors.Example:- "I love dogs" and "I like puppies" → vectors will be **close**- "I love dogs" and "The stock market crashed" → vectors will be **far apart**We will use the `sentence-transformers` library, which uses free Hugging Face models.

### Exercise 2: Compare with `all-mpnet-base-v2`

```python
mpnet = SentenceTransformer("all-mpnet-base-v2")
mp_doc_embeddings = mpnet.encode(documents)

def compare_models(query, top_k=2):
    print(f"Query: {query}")
    for name, m, de in [("MiniLM", model, doc_embeddings), ("mpnet", mpnet, mp_doc_embeddings)]:
        scores = cosine_similarity(m.encode([query]), de)[0]
        print(f"  [{name}]")
        for idx in scores.argsort()[::-1][:top_k]:
            print(f"    {scores[idx]:.3f} | {documents[idx]}")

compare_models("Which sport event is coming up?")
compare_models("renewable energy")
```

### Exercise 3: Simple FAQ bot (5 Q&A pairs)

```python
faq = {
    "How do I reset my password?": "Click 'Forgot password' on the login page and follow the email link.",
    "What are your working hours?": "We are open Monday to Friday, 9 AM to 6 PM.",
    "How can I get a refund?": "Refunds can be requested within 30 days from your orders page.",
    "Do you ship internationally?": "Yes, we ship to over 50 countries.",
    "How do I contact support?": "Email support@example.com or use the live chat.",
}
faq_questions = list(faq.keys())
faq_embeddings = model.encode(faq_questions)

def faq_bot(user_question):
    scores = cosine_similarity(model.encode([user_question]), faq_embeddings)[0]
    best = scores.argmax()
    print(f"You: {user_question}")
    print(f"Matched: {faq_questions[best]} (score {scores[best]:.3f})")
    print(f"Bot: {faq[faq_questions[best]]}\n")

faq_bot("I forgot my login password")
faq_bot("When are you open?")
faq_bot("Can I send a parcel abroad?")
```
